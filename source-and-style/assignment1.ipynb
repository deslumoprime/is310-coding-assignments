# **Assignment 1**

For this assignment, I will be focusing on That Pitch. This serves as a digital library through sync licensing. 

The TLDR is that an artist will upload their work, and That Sync will organize them into libraries. These libraries of music can then be used by filmmakers and game devs. 

Normally, when artists make music *for* a film or game, they're paid to do that job and only that. This is compared to when artists make a song and put it onto Spotify, where they'll get paid the better their song does. 

Sync licensing allows artists to get paid for placement AND for doing well. Artists retain ownership of songs, and people who want to "place" that copyrighted song in their work can still do so.
